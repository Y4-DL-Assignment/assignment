```python
import matplotlib.pyplot as plt
from deep_ql.dql_snake import train_agent, SnakeGame, GRID_SIZE, DQNAgent

env = SnakeGame(grid_size=GRID_SIZE)
agent = DQNAgent()

scores, rewards, epsilons = train_agent(env, agent, episodes=12000)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.plot(scores)
plt.title('Score per Episode')
plt.xlabel('Episode')
plt.ylabel('Score')

plt.subplot(1,3,2)
plt.plot(rewards)
plt.title('Total Reward per Episode')
plt.xlabel('Episode')
plt.ylabel('Reward')

plt.subplot(1,3,3)
plt.plot(epsilons)
plt.title('Epsilon Decay')
plt.xlabel('Episode')
plt.ylabel('Epsilon')

plt.tight_layout()
plt.show()
```
