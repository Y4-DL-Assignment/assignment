# 🐍 Deep SARSA – Snake Game Training
This notebook trains a Deep SARSA agent on the Snake environment and visualizes:
- Score per Episode  
- Total Reward per Episode  
- Epsilon Decay Curve


```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import importlib, snake_deep_sarsa
importlib.reload(snake_deep_sarsa)
from snake_deep_sarsa import SnakeGame, SARSAAgent, train_agent_sarsa, GRID_SIZE
```

```python
env = SnakeGame(grid_size=GRID_SIZE)
agent = SARSAAgent()
```

```python
episodes = 20000  # same as your DQN example

scores, rewards, epsilons = train_agent_sarsa(env, agent, episodes)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(scores)
plt.title('Score per Episode')
plt.xlabel('Episode')
plt.ylabel('Score')

plt.subplot(1, 3, 2)
plt.plot(rewards)
plt.title('Total Reward per Episode')
plt.xlabel('Episode')
plt.ylabel('Reward')

plt.subplot(1, 3, 3)
plt.plot(epsilons)
plt.title('Epsilon Decay')
plt.xlabel('Episode')
plt.ylabel('Epsilon')

plt.tight_layout()
plt.show()

print(f"\nTraining complete! Best score: {max(scores)}")
print(f"Final 100-episode average: {np.mean(scores[-100:]):.2f}")
```
